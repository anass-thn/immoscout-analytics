```python
#compter le nbr de ventes par types de bien
dvf['Type local'].value_counts(dropna=False)
```

```python
# 1. on conserve (maisons et apparts)
dvf_logements = dvf[dvf['Type local'].isin(['Maison', 'Appartement'])]
```

```python
# 2.  affiche les lignes restantes 
print("nombre de transaction d'habitations :",len(dvf_logements))
```

```python
# Combien d'habitations ont un prix manquant ou à 0 ?
dvf_logements['Valeur fonciere'] = (
    dvf_logements['Valeur fonciere']
    .astype(str)
    .str.replace(',', '.', regex=False)
)

dvf_logements['Valeur fonciere'] = pd.to_numeric(
    dvf_logements['Valeur fonciere'],
    errors='coerce'
)
prix_invalides = dvf_logements['Valeur fonciere'].isna() | (dvf_logements['Valeur fonciere'] <= 0)

# Combien d'habitations ont une surface manquante ou à 0 ?
dvf_logements['Surface reelle bati'] = (
    dvf_logements['Surface reelle bati']
    .astype(str)
    .str.replace(',', '.', regex=False)
)

dvf_logements['Surface reelle bati'] = pd.to_numeric(
    dvf_logements['Surface reelle bati'],
    errors='coerce'
)

surfaces_invalides = dvf_logements['Surface reelle bati'].isna() | (dvf_logements['Surface reelle bati'] <= 0)

print("Logements sans prix valide :", prix_invalides.sum())
print("Logements sans surface valide :", surfaces_invalides.sum())
```

```python
# nettoyage des donnees
dvf_clean = dvf_logements[
    (dvf_logements['Valeur fonciere'] > 0) &
    (dvf_logements['Surface reelle bati'] > 0)
]

print(len(dvf_clean))
```

```python
#Calculer le Prix au m^2
dvf_clean['prix_m2'] = dvf_clean['Valeur fonciere'] / dvf_clean['Surface reelle bati']
print(dvf_clean['prix_m2'].describe())


```

```python
pd.options.display.float_format = '{:.2f}'.format
```

```python
# filtrage sur une tranche entre 100 et 30000
dvf_filtres = dvf_clean[
    (dvf_clean['prix_m2'] >= 100) & (dvf_clean['prix_m2'] <= 30000 )
]

print(dvf_filtres['prix_m2'].describe())
```
