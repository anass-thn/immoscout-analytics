```python
#compter le nbr de ventes par types de bien
dvf['Type local'].value_counts(dropna=False)
```

```python
# 1. on conserve (maisons et apparts)
dvf_logements = dvf[dvf['Type local'].isin(['Maison', 'Appartement'])]
```

```python
# 2.  affiche les lignes restantes 
print("nombre de transaction d'habitations :",len(dvf_logements))
```

```python
# Combien d'habitations ont un prix manquant ou à 0 ?
dvf_logements['Valeur fonciere'] = (
    dvf_logements['Valeur fonciere']
    .astype(str)
    .str.replace(',', '.', regex=False)
)

dvf_logements['Valeur fonciere'] = pd.to_numeric(
    dvf_logements['Valeur fonciere'],
    errors='coerce'
)
prix_invalides = dvf_logements['Valeur fonciere'].isna() | (dvf_logements['Valeur fonciere'] <= 0)

# Combien d'habitations ont une surface manquante ou à 0 ?
dvf_logements['Surface reelle bati'] = (
    dvf_logements['Surface reelle bati']
    .astype(str)
    .str.replace(',', '.', regex=False)
)

dvf_logements['Surface reelle bati'] = pd.to_numeric(
    dvf_logements['Surface reelle bati'],
    errors='coerce'
)

surfaces_invalides = dvf_logements['Surface reelle bati'].isna() | (dvf_logements['Surface reelle bati'] <= 0)

print("Logements sans prix valide :", prix_invalides.sum())
print("Logements sans surface valide :", surfaces_invalides.sum())
```

```python
# nettoyage des donnees
dvf_clean = dvf_logements[
    (dvf_logements['Valeur fonciere'] > 0) &
    (dvf_logements['Surface reelle bati'] > 0)
]

print(len(dvf_clean))
```

```python
# supprimer les duplicant
dvf_clean = dvf_clean.drop_duplicates(
    subset=['Date mutation','Valeur fonciere','Code departement','Code commune','Voie'], keep='first'
)
print("Apres dedupliaction:",len(dvf_clean))
```

```python
#Calculer le Prix au m^2
dvf_clean['prix_m2'] = dvf_clean['Valeur fonciere'] / dvf_clean['Surface reelle bati']
print(dvf_clean['prix_m2'].describe())


```

```python
pd.options.display.float_format = '{:.2f}'.format
```

```python
# filtrage sur une tranche entre 100 et 30000
dvf_filtres = dvf_clean[
    (dvf_clean['prix_m2'] >= 100) & (dvf_clean['prix_m2'] <= 30000 )
]

print(dvf_filtres['prix_m2'].describe())
```

```python
# la reconstruction du code INSEE dans tableau dvf
dvf_filtres['code_insee'] = (
    dvf_filtres['Code departement'].astype(str).str.zfill(2) + 
    dvf_filtres['Code commune'].astype(str).str.zfill(3)
 )
dvf_filtres[['Code departement','Code commune', 'code_insee']].head()
```

```python
# preparer la cle dans la table des loyers
loyers['code_insee'] = loyers['INSEE_C'].astype(str).str.zfill(5)
```

```python
# Fusion entre deux tableaux
dvf_enrichi = pd.merge(
    dvf_filtres,
    loyers[['code_insee', 'loypredm2']],
    on='code_insee',
    how='inner'
)

print("Nombre de lignes après jointure :", len(dvf_enrichi))
dvf_enrichi[['code_insee', 'prix_m2', 'loypredm2']].head()
```

```python
# Convertir loypredm2 en vrai nombre (float)
dvf_enrichi['loypredm2'] = dvf_enrichi['loypredm2'].astype(str).str.replace(',','.',regex=False)
dvf_enrichi['loypredm2'] = pd.to_numeric(dvf_enrichi['loypredm2'], errors='coerce')
```

```python
# Calculer le Rendement Locatif Brut (%)
dvf_enrichi['rendement_brut'] = (dvf_enrichi['loypredm2'] * 12 / dvf_enrichi['prix_m2'])*100
print(dvf_enrichi['rendement_brut'].describe())
```

Enquêter sur les "fausses pépites"  
affichons les 5 biens ayant le rendement le plus élevé

```python
dvf_enrichi = dvf_enrichi.sort_values(by='rendement_brut',ascending=False)
dvf_enrichi[['Commune', 'Type local', 'Valeur fonciere', 'Surface reelle bati', 'prix_m2', 'loypredm2', 'rendement_brut']].head(5)
```

```python
# calcul d'une médiane par groupe :
dvf_enrichi['prix_m2_median_commune'] = dvf_enrichi.groupby(['code_insee', 'Type local'])['prix_m2'].transform('median')
dvf_enrichi[['code_insee','Commune','prix_m2','prix_m2_median_commune']]
```

```python
# Compter le nombre de ventes par commune et type de bien
dvf_enrichi['nb_ventes_commune_type'] = dvf_enrichi.groupby(['code_insee', 'Type local'])['prix_m2'].transform('count')

# Regardons à quoi ressemble la distribution du nombre de ventes
print(dvf_enrichi['nb_ventes_commune_type'].describe())
```

```python
# Calcul d'une remise ou décote en % 
dvf_enrichi['decote_pct'] = ((dvf_enrichi['prix_m2_median_commune'] - dvf_enrichi['prix_m2']) / dvf_enrichi['prix_m2_median_commune']) * 100
dvf_enrichi[['code_insee','Commune','prix_m2','prix_m2_median_commune','decote_pct']].head()
```

 Filtrer et isoler les pépites
  créer un nouveau DataFrame appelé pepites

```python
pepites = dvf_enrichi[
    (dvf_enrichi['decote_pct']>=15) & (dvf_enrichi['decote_pct']<=40)
     & (dvf_enrichi['rendement_brut']>=7) & (dvf_enrichi['rendement_brut']<=15) 
     & (dvf_enrichi['Surface reelle bati']>=15) & (dvf_enrichi['nb_ventes_commune_type'] >= 10)

]
print("pepites trouvees :", len(pepites))
pepites [['Commune', 'Type local', 'Surface reelle bati', 'Valeur fonciere', 'prix_m2', 'prix_m2_median_commune', 'decote_pct', 'rendement_brut']]
```

sauvegarder DataFrame pepites dans un fichier CSV, on utilise la méthode .to_csv().

```python
colonnes_utiles = ['Commune', 'Code postal', 'Code departement', 'code_insee' , 'Date mutation', 'Type local', 'Surface reelle bati', 'Nombre pieces principales' , 'Valeur fonciere', 'prix_m2', 'prix_m2_median_commune', 'decote_pct', 'rendement_brut','nb_ventes_commune_type']
pepites[colonnes_utiles].to_csv(r"C:\Users\anass\Desktop\immoscout-analytics\data\processed\resultat.csv", index=False)
```
